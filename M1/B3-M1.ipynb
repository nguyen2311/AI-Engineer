```python
from transformers import pipeline
```

```python
from transformers import(
   AutoTokenizer,
   AutoModelForSequenceClassification,
   pipeline
)
model_id = "niru-nny/SMS_Spam_Detection"
tokenizner = AutoTokenizer.from_pretrained(
    model_id,force_download=True
)
model = AutoModelForSequenceClassification.from_pretrained(
    model_id,force_download=True
)
classifier = pipeline(
    "text-classification",
    model=model,
    tokenizer=tokenizer,
    device = -1
)
tests = [
    "Hey, are we still meeting for lunch?",
    "Please send me the lecture notes.",
    "URGENT! You won a FREE iPhone. Claim your prize now!",
    "Congratulations! Call this number to claim your cash reward.",
    "Your appointment is confirmed for tomorrow."
]
for msg in tests:
    print(msg)
    print(classifier(msg, top_k=None))
    print("-" * 50)
print("Đã khởi tạo mô hình!" if classifier is not None else "Chưa hoàn thành TODO")
```

```python
message = "Congratulations! You won a free prize. Call now!"
result= classifier(message)
print(result)
```

```python
messages = [
    "Hi, are we meeting at 5 pm today?",
    "WIN CASH NOW! Claim your free gift today!",
    "Please send me your homework."
]
results = classifier(messages)

for message, prediction in zip(messages, results):
    print(message)
    print(prediction["label"],round(prediction["score"],3))
    print("---")
```

```python
assert callable(classifier), "Chưa tải mô hình ở bước 3"
assert isinstance(result, list) and len(result) == 1, "Chưa dự đoán ở bước 4"
assert len(results) == len(messages), "Chưa dự đoán đủ danh sách ở bước 5"
assert all("label" in x and "score" in x for x in results), "Kết quả chưa đúng định dạng"
print("✅ Hoàn thành bài thực hành!")
```
